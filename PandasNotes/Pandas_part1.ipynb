Selection in Pandas--> by iloc & loc or we can use python native approach

```python
#iloc--> index based

#creating dataset
df=pd.DataFrame({
    "Name":['Ali','Hamza','Nimrah'],
    "CGPA":[3.91,3.95,3.80],
    "Age":[20,19,20]

})
```

```python
print(df)
```

```python
df.iloc[0,0]
```

```python
df.iloc[2,1]
```
